```python
import pandas as pd
import numpy as np
```

# Creating a dataframe with data

```python
stone_size_arr = [] # S
treatment_arr = []  # T
y_arr = []          # Y

# add 81 examples of T=A, S=s and Y=1
for i in range(0, 81):
    stone_size_arr.append('s')
    treatment_arr.append('A')
    y_arr.append(1)

# add 6 examples of T=A, S=s and Y=0
for i in range(0, 6):
    stone_size_arr.append('s')
    treatment_arr.append('A')
    y_arr.append(0)

# add 234 examples of T=B, S=s and Y=1
for i in range(0, 234):
    stone_size_arr.append('s')
    treatment_arr.append('B')
    y_arr.append(1)

# add 36 examples of T=B, S=s and Y=0
for i in range(0, 36):
    stone_size_arr.append('s')
    treatment_arr.append('B')
    y_arr.append(0)

# add 192 examples of T=A, S=l and Y=1
for i in range(0, 192):
    stone_size_arr.append('l')
    treatment_arr.append('A')
    y_arr.append(1)

# add 71 examples of T=A, S=l and Y=0
for i in range(0, 71):
    stone_size_arr.append('l')
    treatment_arr.append('A')
    y_arr.append(0)

# add 55 examples of T=B, S=l and Y=1
for i in range(0, 55):
    stone_size_arr.append('l')
    treatment_arr.append('B')
    y_arr.append(1)
    pass

# add 25 examples of T=B, S=l and Y=0
for i in range(0, 25):
    stone_size_arr.append('l')
    treatment_arr.append('B')
    y_arr.append(0)

df = pd.DataFrame({})
df['S'] = stone_size_arr
df['T'] = treatment_arr
df['Y'] = y_arr
```

# Small stones

```python
df_tmp = df[df['S'] == 's']
len(df_tmp)
```

```python
df_tmp_A = df_tmp[df_tmp['T'] == 'A']
print('Treatment A: {} / {} = {}'.format(
    df_tmp_A['Y'].sum(), len(df_tmp_A), np.round(df_tmp_A['Y'].mean(), 2)
))
```

```python
df_tmp_B = df_tmp[df_tmp['T'] == 'B']
print('Treatment B: {} / {} = {}'.format(
    df_tmp_B['Y'].sum(), len(df_tmp_B), np.round(df_tmp_B['Y'].mean(), 2)
))
```

<font size="8">Back-Door adjustments</font>

```python
df_S_s = df[df['S'] == 's']
df_S_l = df[df['S'] == 'l']
p_S_s = len(df_S_s) / len(df)
p_S_l = len(df_S_l) / len(df)

print('Probability S=s is {}'.format(p_S_s))
print('Probability S=l is {}'.format(p_S_l))

Y_A = df_S_s[df_S_s['T'] == 'A'].Y.mean() * p_S_s + df_S_l[df_S_l['T'] == 'A'].Y.mean() * p_S_l
Y_B = df_S_s[df_S_s['T'] == 'B'].Y.mean() * p_S_s + df_S_l[df_S_l['T'] == 'B'].Y.mean() * p_S_l

ate = Y_A - Y_B

print('Treatmen A: {}'.format(np.round(Y_A, 2)))
print('Treatmen B: {}'.format(np.round(Y_B, 2)))
print('ATE: {}'.format(np.round(ate, 2)))
```
